### Notebook that takes a dataset and creates a smaller sample and adds a column to the data frame for potential fraud weights

```python
import pandas as pd

df = pd.read_csv("small_sample.csv")
```

```python
def calculate_column3(row):
    if row['icr_or_manifest'] == 1 and row['symbology_mismatch'] == 1:
        return 1
    elif (row['icr_or_manifest'] == 1 and row['symbology_mismatch'] == 2) or (row['icr_or_manifest'] == 2 and row['symbology_mismatch'] == 1):
        return 2
    elif row['icr_or_manifest'] == 2 and row['symbology_mismatch'] == 2:
        return 4
    else:
        return None  # or any default value
```

```python
df['attaches_to_weight'] = df.apply(calculate_column3, axis=1)
```

```python
df[['symbology_mismatch', 'icr_or_manifest', 'attaches_to_weight']].value_counts()
```

```python
df.to_csv('small_sample.csv', index=False)
```
